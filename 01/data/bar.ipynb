```python
import matplotlib.pyplot as plt
figure,axes=plt.subplots(nrows=1, ncols=2,figsize=(15,6),dpi=100)
countries=["China","US","UK","Japan","Korean","Russia","France"]
values=[35,23,18,21,56,78,51]
axes[0].bar(countries,values,width=0.5,color="green")
axes[0].set_title("oil",fontsize=20)
axes[0].set_xlabel("countries",fontsize=20)
axes[0].set_ylabel("oil values")
axes[0].grid(linestyle="--",alpha=0.3)
countries2=["China","US","UK","Japan","Korean","Russia","France"]
values2=[14,14,3,3,3,2,2]
axes[1].pie(values2,labels=countries2,autopct='%1.2f%%')
axes[1].set_title("water",fontsize=20)
axes[1].legend(loc="lower center",ncol=5,bbox_to_anchor=(0.5, -0.15))
plt.savefig("01.png")

```
